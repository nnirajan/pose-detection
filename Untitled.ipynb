```python
import cv2
import mediapipe as mp
import numpy as np
from enum import Enum, auto

# Initialize MediaPipe
mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose

def draw_text_with_bg(img, text, org, font, font_scale, text_color, bg_color, thickness=2, pad_x=10, pad_y=5):
    """Draw text with background and padding"""
    (text_w, text_h), baseline = cv2.getTextSize(text, font, font_scale, thickness)
    x, y = org
    cv2.rectangle(img, 
                 (x - pad_x, y - text_h - pad_y), 
                 (x + text_w + pad_x, y + baseline + pad_y), 
                 bg_color, cv2.FILLED)
    cv2.putText(img, text, (x, y), font, font_scale, text_color, thickness, cv2.LINE_AA)

class ExerciseType(Enum):
    BICEP_CURL = auto()
    SQUAT = auto()

class ExerciseConfig:
    def __init__(self, exercise_type):
        self.exercise_type = exercise_type
        self.MIN_CONFIDENCE = 0.8
        
        if self.exercise_type == ExerciseType.BICEP_CURL:
            self.ANGLE_THRESHOLDS = (60, 160)
            self.FORM_GUIDANCE = "FULL RANGE NEEDED!"
            self.COUNTER_TEXT = "Bicep Curls"
            self.ANGLE_TEXT = "Elbow Angle"
            self.SWITCHABLE_SIDES = True
        else:  # SQUAT
            self.ANGLE_THRESHOLDS = (70, 160)
            self.FORM_GUIDANCE = "GO DEEPER!"
            self.COUNTER_TEXT = "Squats"
            self.ANGLE_TEXT = "Knee Angle"
            self.SWITCHABLE_SIDES = False

class ExerciseCounter:
    def __init__(self, config):
        self.config = config
        self.pose = mp_pose.Pose(
            min_detection_confidence=config.MIN_CONFIDENCE,
            min_tracking_confidence=config.MIN_CONFIDENCE
        )
        self.reset()

    def reset(self):
        self.count = 0
        self.state = "up"
        self.current_rom = 0

    def calculate_angle(self, a, b, c):
        a, b, c = np.array(a), np.array(b), np.array(c)
        ba = a - b
        bc = c - b
        norm_ba = np.linalg.norm(ba)
        norm_bc = np.linalg.norm(bc)
        if norm_ba == 0 or norm_bc == 0:
            return None
        cosine_angle = np.dot(ba, bc) / (norm_ba * norm_bc)
        cosine_angle = np.clip(cosine_angle, -1.0, 1.0)
        return np.degrees(np.arccos(cosine_angle))

    def process_bicep_curl(self, landmarks, side):
        shoulder = landmarks[getattr(mp_pose.PoseLandmark, f"{side.upper()}_SHOULDER")]
        elbow = landmarks[getattr(mp_pose.PoseLandmark, f"{side.upper()}_ELBOW")]
        wrist = landmarks[getattr(mp_pose.PoseLandmark, f"{side.upper()}_WRIST")]
        
        angle = self.calculate_angle(
            [shoulder.x, shoulder.y],
            [elbow.x, elbow.y],
            [wrist.x, wrist.y]
        )
        
        if angle is not None:
            if angle > self.config.ANGLE_THRESHOLDS[1]:
                self.state = "down"
            elif angle < self.config.ANGLE_THRESHOLDS[0] and self.state == "down":
                self.state = "up"
                self.count += 1
        return angle, None

    def process_squat(self, landmarks):
        # Left leg
        hip_l = landmarks[mp_pose.PoseLandmark.LEFT_HIP]
        knee_l = landmarks[mp_pose.PoseLandmark.LEFT_KNEE]
        ankle_l = landmarks[mp_pose.PoseLandmark.LEFT_ANKLE]
        
        # Right leg
        hip_r = landmarks[mp_pose.PoseLandmark.RIGHT_HIP]
        knee_r = landmarks[mp_pose.PoseLandmark.RIGHT_KNEE]
        ankle_r = landmarks[mp_pose.PoseLandmark.RIGHT_ANKLE]

        angle_l = self.calculate_angle(
            [hip_l.x, hip_l.y],
            [knee_l.x, knee_l.y],
            [ankle_l.x, ankle_l.y]
        )
        angle_r = self.calculate_angle(
            [hip_r.x, hip_r.y],
            [knee_r.x, knee_r.y],
            [ankle_r.x, ankle_r.y]
        )

        angle = (angle_l + angle_r)/2 if angle_l and angle_r else None

        if angle is not None:
            if angle < self.config.ANGLE_THRESHOLDS[0] and self.state == "up":
                self.state = "down"
                self.count += 1
            elif angle > self.config.ANGLE_THRESHOLDS[1] and self.state == "down":
                self.state = "up"
        return angle, (angle_l, angle_r)

    def process_frame(self, landmarks, side='right'):
        if self.config.exercise_type == ExerciseType.BICEP_CURL:
            return self.process_bicep_curl(landmarks, side)
        else:
            return self.process_squat(landmarks)

class FitnessTrainer:
    def __init__(self):
        self.current_exercise = ExerciseType.BICEP_CURL
        self.config = ExerciseConfig(self.current_exercise)
        self.counter = ExerciseCounter(self.config)
        self.use_right_side = True
        self.cap = cv2.VideoCapture(0)
        self.current_angle = None
        self.angle_details = None

    def switch_exercise(self):
        self.current_exercise = (
            ExerciseType.SQUAT if self.current_exercise == ExerciseType.BICEP_CURL 
            else ExerciseType.BICEP_CURL
        )
        self.config = ExerciseConfig(self.current_exercise)
        self.counter = ExerciseCounter(self.config)
        self.use_right_side = True

    def run(self):
        while self.cap.isOpened():
            success, frame = self.cap.read()
            if not success:
                continue

            display_frame = cv2.flip(frame, 1)
            h, w = display_frame.shape[:2]

            # Process frame
            results = self.counter.pose.process(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

            if results.pose_landmarks:
                # Draw skeleton
                annotated_frame = frame.copy()
                mp_drawing.draw_landmarks(annotated_frame, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)
                display_frame = cv2.flip(annotated_frame, 1)
                
                # Process exercise
                side = 'right' if self.use_right_side else 'left'
                self.current_angle, self.angle_details = self.counter.process_frame(
                    results.pose_landmarks.landmark, 
                    side
                )

                # Draw exercise-specific visuals
                if self.config.exercise_type == ExerciseType.BICEP_CURL:
                    self._draw_bicep_ui(display_frame, results, w, h)
                else:
                    self._draw_squat_ui(display_frame, results, w, h)

            # Common UI
            self._draw_common_ui(display_frame, h, w)

            cv2.imshow('Fitness Trainer', display_frame)
            
            key = cv2.waitKey(1)
            if key == ord('q'):
                break
            elif key == ord('e'):
                self.switch_exercise()
            elif key == ord('r'):
                self.counter.reset()
            elif key == ord('s') and self.config.SWITCHABLE_SIDES:
                self.use_right_side = not self.use_right_side
                self.counter.reset()

        self.cap.release()
        cv2.destroyAllWindows()

    def _draw_bicep_ui(self, frame, results, w, h):
        if self.current_angle is not None:
            side = 'right' if self.use_right_side else 'left'
            elbow = results.pose_landmarks.landmark[getattr(mp_pose.PoseLandmark, f"{side.upper()}_ELBOW")]
            elbow_pos = (int((1 - elbow.x) * w), int(elbow.y * h))
            
            # Draw angle arc
            cv2.ellipse(frame, elbow_pos, (30,30), 0, 0, int(self.current_angle), (0,255,0), 2)
            cv2.putText(frame, f"{int(self.current_angle)}°", 
                        (elbow_pos[0]+40, elbow_pos[1]), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

    def _draw_squat_ui(self, frame, results, w, h):
        if self.angle_details:
            angle_l, angle_r = self.angle_details
            # Left knee
            knee_l = results.pose_landmarks.landmark[mp_pose.PoseLandmark.LEFT_KNEE]
            knee_pos_l = (int((1 - knee_l.x) * w), int(knee_l.y * h))
            if angle_l:
                cv2.ellipse(frame, knee_pos_l, (30,30), 0, 0, int(angle_l), (0,255,0), 2)
                cv2.putText(frame, f"{int(angle_l)}°", 
                            (knee_pos_l[0]+40, knee_pos_l[1]), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)
            
            # Right knee
            knee_r = results.pose_landmarks.landmark[mp_pose.PoseLandmark.RIGHT_KNEE]
            knee_pos_r = (int((1 - knee_r.x) * w), int(knee_r.y * h))
            if angle_r:
                cv2.ellipse(frame, knee_pos_r, (30,30), 0, 0, int(angle_r), (0,255,0), 2)
                cv2.putText(frame, f"{int(angle_r)}°", 
                            (knee_pos_r[0]+40, knee_pos_r[1]), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,255,0), 2)

    def _draw_common_ui(self, frame, h, w):
        # Exercise info
        exercise_name = "Bicep Curl" if self.current_exercise == ExerciseType.BICEP_CURL else "Squat"
        draw_text_with_bg(frame, f"Exercise: {exercise_name}", 
                         (w-300, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,0), (255,255,255))
        
        # Main metrics
        angle_text = f"{self.config.ANGLE_TEXT}: {int(self.current_angle) if self.current_angle else '--'}°"
        draw_text_with_bg(frame, angle_text,
                         (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,0,0), (255,255,255))
        
        draw_text_with_bg(frame, f"{self.config.COUNTER_TEXT}: {self.counter.count}", 
                         (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0,0,0), (255,255,255))
        
        # Thresholds
        thresholds_text = f"Thresholds: {self.config.ANGLE_THRESHOLDS[0]}°-{self.config.ANGLE_THRESHOLDS[1]}°"
        draw_text_with_bg(frame, thresholds_text,
                         (w-400, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,0), (255,255,255))
        
        # Form guidance
        if (self.current_angle and 
            self.config.ANGLE_THRESHOLDS[0] < self.current_angle < self.config.ANGLE_THRESHOLDS[1]):
            draw_text_with_bg(frame, self.config.FORM_GUIDANCE,
                            (w//4, h-40), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0,0,255), (255,255,255))
        
        # Controls
        controls_text = "[E] Switch Exercise | "
        if self.config.SWITCHABLE_SIDES:
            controls_text += "[S] Switch Side | "
        controls_text += "[R] Reset | [Q] Quit"
        
        draw_text_with_bg(frame, controls_text,
                         (10, h-20), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,0,0), (255,255,255))

# Main Execution
if __name__ == "__main__":
    trainer = FitnessTrainer()
    trainer.run()
```
